## 1. Algebra and Functions

*Đại số và hàm số*

Solve $e^{2x}=7$ for $x$, solve the inequality $x^2-5x+6<0$, and simplify $(x^2-1)/(x-1)$. For $y=x^2+1$, find the possible values of $x$ in terms of $y$. A logarithm with base $b$ is defined by $\log_b(a)=\frac{\ln(a)}{\ln(b)}$.

*Giải $e^{2x}=7$ theo $x$, giải bất phương trình $x^2-5x+6<0$ và rút gọn $(x^2-1)/(x-1)$. Với $y=x^2+1$, tìm các giá trị có thể có của $x$ theo $y$. Logarit cơ số $b$ được xác định bởi $\log_b(a)=\frac{\ln(a)}{\ln(b)}$.*

```python
equation_roots = sp.solve(sp.Eq(sp.exp(2*x), 7), x)
inequality = sp.solve_univariate_inequality(x**2 - 5*x + 6 < 0, x)
simplified = sp.simplify((x**2 - 1) / (x - 1))
inverse_branches = sp.solve(sp.Eq(y, x**2 + 1), x)
print('Solutions to e^(2x)=7 | Nghiệm e^(2x)=7:', equation_roots)
print('Solution set for x^2-5x+6<0 | Miền nghiệm x^2-5x+6<0:', inequality)
print('Simplified | Rút gọn:', simplified, '| nghiệm ngược:', inverse_branches)
print('log_2(8) =', math.log(8, 2))
```

## 2. Linear Algebra

*Đại số tuyến tính*

Let $A=\begin{bmatrix}1&2&3\\0&1&4\\5&6&0\end{bmatrix}$ and $B=\begin{bmatrix}2&0&1\\1&4&2\\0&0&1\end{bmatrix}$. Common operations include $\det(A)$, $A^{-1}$, $AB$, $A^T$, $\operatorname{rank}(A)$, $\operatorname{tr}(A)$, and solving $A\mathbf{x}=\mathbf{b}$. Reduced row-echelon form identifies pivot columns; eigenvalues satisfy $\det(A-\lambda I)=0$.

*Cho $A=\begin{bmatrix}1&2&3\\0&1&4\\5&6&0\end{bmatrix}$ và $B=\begin{bmatrix}2&0&1\\1&4&2\\0&0&1\end{bmatrix}$. Các phép toán thường dùng gồm định thức, ma trận nghịch đảo, tích và chuyển vị ma trận, hạng, vết, và giải $A\mathbf{x}=\mathbf{b}$. Dạng bậc thang rút gọn xác định các cột trụ; trị riêng thỏa $\det(A-\lambda I)=0$.*

```python
A = sp.Matrix([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
B = sp.Matrix([[2, 0, 1], [1, 4, 2], [0, 0, 1]])
print('det(A) =', A.det())
print('Inverse of A | A nghịch đảo =\n', A.inv())
print('A B =\n', A * B)
print('Transpose of A | A chuyển vị =\n', A.T)
print('rank(A), trace(A) =', A.rank(), A.trace())
print('Solution to A v | Nghiệm A v = (1,2,3):', A.LUsolve(sp.Matrix([1, 2, 3])))
print('RREF and pivot columns | RREF và cột trụ:', A.rref())
print('Characteristic polynomial | Đa thức đặc trưng:', sp.factor(A.charpoly().as_expr()))
print('Eigenvalues and eigenvectors | Trị riêng và vector riêng:', A.eigenvects())
print('det(A+B) =', (A + B).det())
```

## 3. Counting and Discrete Probability

*Tổ hợp và xác suất rời rạc*

For $X\sim\operatorname{Bin}(n,p)$, $P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}$. For $X\sim\operatorname{Poisson}(\lambda)$, $P(X=k)=e^{-\lambda}\lambda^k/k!$. Bayes theorem gives $P(H\mid E)=P(E\mid H)P(H)/P(E)$. If $X$ takes values $x_i$ with probabilities $p_i$, then $E[X]=\sum_i x_ip_i$ and $\operatorname{Var}(X)=\sum_i p_i(x_i-E[X])^2$.

Example: let $n=10$ and $p=1/4$; find $P(X=3)$ and $P(X\leq3)$. For a Poisson distribution with $\lambda=3$, find $P(X=2)$.

*Với $X\sim\operatorname{Bin}(n,p)$, $P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}$. Với $X\sim\operatorname{Poisson}(\lambda)$, $P(X=k)=e^{-\lambda}\lambda^k/k!$. Định lý Bayes cho $P(H\mid E)=P(E\mid H)P(H)/P(E)$. Nếu $X$ nhận các giá trị $x_i$ với xác suất $p_i$, thì $E[X]=\sum_i x_ip_i$ và $\operatorname{Var}(X)=\sum_i p_i(x_i-E[X])^2$.

Ví dụ: cho $n=10$, $p=1/4$, hãy tính $P(X=3)$ và $P(X\leq3)$. Với phân phối Poisson có $\lambda=3$, hãy tính $P(X=2)$.*

```python
n_trials, k, p = 10, 3, sp.Rational(1, 4)
binomial_exact = sp.binomial(n_trials, k) * p**k * (1-p)**(n_trials-k)
print('Exact P(Bin(10,1/4)=3) | Xác suất chính xác =', binomial_exact)
print('P(X<=3) | Xác suất tích lũy =', stats.binom.cdf(k, n_trials, float(p)))
print('P(Poisson(3)=2) | Xác suất Poisson =', stats.poisson.pmf(2, mu=3))
prior, evidence_given_h, evidence_given_not_h = 0.01, 0.9, 0.05
posterior = evidence_given_h * prior / (evidence_given_h*prior + evidence_given_not_h*(1-prior))
print('Bayes P(H|E) | Xác suất Bayes =', posterior)
values, probabilities = np.array([1, 2, 3]), np.array([0.2, 0.5, 0.3])
expected = np.dot(values, probabilities)
variance = np.dot((values-expected)**2, probabilities)
print('E[X], Var(X) | Kỳ vọng, phương sai =', expected, variance)
```

## 4. Continuous Distributions

*Phân phối liên tục*

If $X\sim N(\mu,\sigma^2)$, then $Z=(X-\mu)/\sigma\sim N(0,1)$ and $P(X\leq a)=\Phi((a-\mu)/\sigma)$. For $X\sim U(a,b)$, $P(c\leq X\leq d)=(d-c)/(b-a)$ after clipping the interval to $[a,b]$. For an exponential distribution with rate $\lambda$, $P(X\leq t)=1-e^{-\lambda t}$.

Example: for $X\sim N(100,15^2)$, find $P(X<115)$; for $U\sim U(0,1)$, find $P(1/4\leq U\leq3/4)$; for an exponential variable with rate $1/2$, find $P(T\leq2)$.

*Nếu $X\sim N(\mu,\sigma^2)$ thì $Z=(X-\mu)/\sigma\sim N(0,1)$ và $P(X\leq a)=\Phi((a-\mu)/\sigma)$. Với $X\sim U(a,b)$, xác suất $P(c\leq X\leq d)=(d-c)/(b-a)$ sau khi giới hạn khoảng vào $[a,b]$. Với phân phối mũ tốc độ $\lambda$, $P(X\leq t)=1-e^{-\lambda t}$.

Ví dụ: với $X\sim N(100,15^2)$, tính $P(X<115)$; với $U\sim U(0,1)$, tính $P(1/4\leq U\leq3/4)$; với biến mũ tốc độ $1/2$, tính $P(T\leq2)$.*

```python
mu, sigma, cutoff = 100, 15, 115
z_score = (cutoff-mu)/sigma
print('z =', z_score, '| P(X<115) =', stats.norm.cdf(cutoff, loc=mu, scale=sigma))
print('P(1/4 <= U <= 3/4) =', stats.uniform.cdf(0.75)-stats.uniform.cdf(0.25))
print('P(T<=2), rate=1/2 =', stats.expon.cdf(2, scale=1/0.5))
print('97.5% quantile of N(0,1) | Phân vị 97.5% của N(0,1) =', stats.norm.ppf(0.975))
```

## 5. Descriptive Statistics

*Thống kê mô tả*

For data $x_1,\dots,x_n$, the mean is $\bar{x}=n^{-1}\sum_i x_i$. Population variance divides by $n$; sample variance divides by $n-1$. Standard deviation is the square root of variance. Sample covariance measures joint variation, and correlation is $r=\operatorname{Cov}(X,Y)/(s_Xs_Y)$.

Example data: $12,15,18,22,30,25,19$.

*Với dữ liệu $x_1,\dots,x_n$, trung bình là $\bar{x}=n^{-1}\sum_i x_i$. Phương sai tổng thể chia cho $n$; phương sai mẫu chia cho $n-1$. Độ lệch chuẩn là căn bậc hai của phương sai. Hiệp phương sai mẫu đo biến thiên đồng thời; hệ số tương quan là $r=\operatorname{Cov}(X,Y)/(s_Xs_Y)$.

Dữ liệu ví dụ: $12,15,18,22,30,25,19$.*

```python
sample = np.array([12, 15, 18, 22, 30, 25, 19], dtype=float)
print('Mean, median | Trung bình, trung vị =', sample.mean(), np.median(sample))
print('Population and sample variance | Phương sai tổng thể và mẫu =', sample.var(ddof=0), sample.var(ddof=1))
print('Sample standard deviation | Độ lệch chuẩn mẫu =', sample.std(ddof=1))
values, counts = np.unique(sample, return_counts=True)
print('Mode =', values[counts == counts.max()].tolist())
X = np.array([1, 2, 3, 4, 5], dtype=float)
Y = np.array([2, 4, 6, 8, 10], dtype=float)
print('Sample Cov(X,Y) | Hiệp phương sai mẫu =', np.cov(X, Y, ddof=1)[0, 1])
print('Corr(X,Y) | Hệ số tương quan =', np.corrcoef(X, Y)[0, 1])
```

## 6. Confidence Intervals and Hypothesis Tests

*Khoảng tin cậy và kiểm định giả thuyết*

When the population standard deviation $\sigma$ is known, a two-sided confidence interval for $\mu$ is $\bar{x}\pm z_{1-\alpha/2}\,\sigma/\sqrt{n}$. When $\sigma$ is unknown, use $\bar{x}\pm t_{1-\alpha/2,n-1}s/\sqrt{n}$. A $z$-test for a mean uses $z=(\bar{x}-\mu_0)/(\sigma/\sqrt{n})$; a one-sample $t$-test uses the sample standard deviation.

Example: use a $95\%$ confidence level and test $H_0:\mu=20$ against a two-sided alternative with the sample above.

*Khi biết độ lệch chuẩn tổng thể $\sigma$, khoảng tin cậy hai phía cho $\mu$ là $\bar{x}\pm z_{1-\alpha/2}\,\sigma/\sqrt{n}$. Khi chưa biết $\sigma$, dùng $\bar{x}\pm t_{1-\alpha/2,n-1}s/\sqrt{n}$. Kiểm định $z$ cho trung bình dùng $z=(\bar{x}-\mu_0)/(\sigma/\sqrt{n})$; kiểm định $t$ một mẫu dùng độ lệch chuẩn mẫu.

Ví dụ: dùng mức tin cậy $95\%$ và kiểm định hai phía $H_0:\mu=20$ với mẫu ở trên.*

```python
confidence = 0.95
alpha = 1-confidence
z_critical = stats.norm.ppf(1-alpha/2)
ci_known_sigma = (sample.mean()-z_critical*4/np.sqrt(len(sample)), sample.mean()+z_critical*4/np.sqrt(len(sample)))
ci_unknown_sigma = stats.t.interval(confidence, df=len(sample)-1, loc=sample.mean(), scale=stats.sem(sample))
t_stat, t_p_value = stats.ttest_1samp(sample, popmean=20)
print('95% CI, known sigma=4 | Khoảng tin cậy 95%, biết sigma=4:', ci_known_sigma)
print('95% CI, unknown sigma | Khoảng tin cậy 95%, chưa biết sigma:', ci_unknown_sigma)
print('Two-sided t-test | Kiểm định t hai phía, H0: mu=20:', t_stat, t_p_value)
z_stat = (103-100)/(10/np.sqrt(100))
print('Two-sided z-test | Kiểm định z hai phía:', z_stat, 2*stats.norm.sf(abs(z_stat)))
```

## 7. Regression and PCA

*Hồi quy và PCA*

Simple linear regression models $Y_i=\beta_0+\beta_1X_i+\varepsilon_i$. Ordinary least squares (OLS) chooses $\hat{\beta}$ to minimize $\sum_i(y_i-\hat{y}_i)^2$. With multiple predictors, $\hat{\beta}=(X^TX)^{-1}X^Ty$ when the inverse exists. PCA uses eigenvectors of the covariance matrix; principal components are ordered by decreasing eigenvalue.

*Hồi quy tuyến tính đơn mô hình hóa $Y_i=\beta_0+\beta_1X_i+\varepsilon_i$. Bình phương tối thiểu thông thường (OLS) chọn $\hat{\beta}$ để cực tiểu hóa $\sum_i(y_i-\hat{y}_i)^2$. Với nhiều biến giải thích, $\hat{\beta}=(X^TX)^{-1}X^Ty$ nếu ma trận khả nghịch. PCA sử dụng vector riêng của ma trận hiệp phương sai; các thành phần chính được sắp theo trị riêng giảm dần.*

```python
reg = stats.linregress(X, Y)
design = np.column_stack((np.ones(len(X)), X))
beta, *_ = np.linalg.lstsq(design, Y, rcond=None)
print('Simple regression | Hồi quy đơn: intercept, slope, r =', reg.intercept, reg.slope, reg.rvalue)
print('OLS [intercept, slope] =', beta)
X_multi = np.array([[1, 0], [0, 1], [2, 1], [1, 2]], dtype=float)
design_multi = np.column_stack((np.ones(len(X_multi)), X_multi))
y_multi = np.array([2, 3, 5, 6], dtype=float)
beta_multi, *_ = np.linalg.lstsq(design_multi, y_multi, rcond=None)
print('Multiple regression OLS | OLS đa biến [intercept, beta1, beta2] =', beta_multi)
X_pca = np.array([[2, 0], [0, 2], [3, 1], [1, 3]], dtype=float)
centered = X_pca-X_pca.mean(axis=0)
eigenvalues, eigenvectors = np.linalg.eigh(np.cov(centered, rowvar=False))
order = np.argsort(eigenvalues)[::-1]
print('PCA explained variance ratio | Tỷ lệ phương sai PCA:', eigenvalues[order]/eigenvalues.sum())
print('Principal component vectors | Vector thành phần chính (theo cột):\n', eigenvectors[:, order])
```

## 8. Differentiation and Single-Variable Optimization

*Vi phân và tối ưu một biến*

Let $f(x)=3x^3-5x^2+2x-7$. The first derivative $f^{\prime}(x)$ measures the rate of change; critical points satisfy $f^{\prime}(x)=0$. For $g(x,y)=3x^2y+y^3$, the partial derivatives are $\partial g/\partial x$ and $\partial g/\partial y$. A standard limit is $\lim_{x\to0}\sin(x)/x=1$. The Taylor expansion around $a$ is $f(x)=\sum_{j=0}^m f^{(j)}(a)(x-a)^j/j!+O((x-a)^{m+1})$.

*Cho $f(x)=3x^3-5x^2+2x-7$. Đạo hàm bậc nhất $f^{\prime}(x)$ mô tả tốc độ biến thiên; điểm tới hạn thỏa $f^{\prime}(x)=0$. Với $g(x,y)=3x^2y+y^3$, các đạo hàm riêng là $\partial g/\partial x$ và $\partial g/\partial y$. Giới hạn cơ bản là $\lim_{x\to0}\sin(x)/x=1$. Khai triển Taylor quanh $a$ là $f(x)=\sum_{j=0}^m f^{(j)}(a)(x-a)^j/j!+O((x-a)^{m+1})$.*

```python
f = 3*x**3-5*x**2+2*x-7
derivative = sp.diff(f, x)
print('First derivative f (x) | Đạo hàm f (x)', derivative, '| f (2) =', derivative.subs(x, 2))
g = 3*x**2*y+y**3
print('Partial derivatives with respect to x and y | Đạo hàm riêng theo x và y =', sp.diff(g, x), sp.diff(g, y))
print('Critical points of f | Điểm tới hạn của f:', sp.solve(derivative, x))
print('lim sin(x)/x as x->0 | Giới hạn sin(x)/x khi x->0 =', sp.limit(sp.sin(x)/x, x, 0))
print('Taylor expansion of sqrt(x) around 4, degree 2 | Khai triển Taylor sqrt(x) quanh 4, bậc 2 =', sp.series(sp.sqrt(x), x, 4, 3).removeO())
print('Expansion of (x+1)^3 | Khai triển (x+1)^3 =', sp.expand((x+1)**3))
print('Factorization of x^2-1 | Phân tích x^2-1 =', sp.factor(x**2-1))
```

## 9. Integration

*Tích phân*

An antiderivative $F$ of $f$ satisfies $F^{\prime}(x)=f(x)$. The definite integral is $\int_a^b f(x)\,dx=F(b)-F(a)$. Examples: evaluate $\int_0^2(3x^3-5x^2+2x-7)\,dx$ and $\int_0^\pi\sin(x)\,dx$. Numerical quadrature provides an approximate check.

*Nguyên hàm $F$ của $f$ thỏa $F^{\prime}(x)=f(x)$. Tích phân xác định là $\int_a^b f(x)\,dx=F(b)-F(a)$. Ví dụ: tính $\int_0^2(3x^3-5x^2+2x-7)\,dx$ và $\int_0^\pi\sin(x)\,dx$. Phép cầu phương số giúp kiểm tra gần đúng.*

```python
print('Antiderivative of exp(-x^2) | Nguyên hàm exp(-x^2) =', sp.integrate(sp.exp(-x**2), x))
print('Polynomial integral from 0 to 2 | Tích phân đa thức từ 0 đến 2 =', sp.integrate(f, (x, 0, 2)))
print('Integral of sin(x) from 0 to pi | Tích phân sin(x) từ 0 đến pi =', sp.integrate(sp.sin(x), (x, 0, sp.pi)))
numeric_value, numeric_error = integrate.quad(lambda value: value**2, 0, 3)
print('Numerical integral of x^2 from 0 to 3 | Tích phân số x^2 từ 0 đến 3 =', numeric_value, '(sai số ước lượng:', numeric_error, ')')
```

## 10. Differential Equations and Constrained Optimization

*Phương trình vi phân và tối ưu có ràng buộc*

The differential equation $y^{\prime}(t)=3y(t)$ has general solution $y(t)=Ce^{3t}$. To optimize $f(x,y)=xy$ subject to $x+y=10$, form the Lagrangian $\mathcal{L}(x,y,\lambda)=xy-\lambda(x+y-10)$ and solve $\nabla_{x,y,\lambda}\mathcal{L}=0$.

*Phương trình vi phân $y^{\prime}(t)=3y(t)$ có nghiệm tổng quát $y(t)=Ce^{3t}$. Để tối ưu $f(x,y)=xy$ với ràng buộc $x+y=10$, lập hàm Lagrange $\mathcal{L}(x,y,\lambda)=xy-\lambda(x+y-10)$ và giải $\nabla_{x,y,\lambda}\mathcal{L}=0$.*

```python
time = sp.symbols('time')
function_y = sp.Function('y')
ode_solution = sp.dsolve(sp.Eq(sp.diff(function_y(time), time), 3*function_y(time)), function_y(time))
print('Solution to y = 3y derivative | Nghiệm y = 3y:', ode_solution)
lam = sp.symbols('lambda')
lagrange_solution = sp.solve([y-lam, x-lam, x+y-10], [x, y, lam], dict=True)
print('Extremum of xy subject to x+y=10 | Cực trị xy với x+y=10:', lagrange_solution)
```

## 11. Number Theory, Modular Arithmetic, and Logic

*Số học, modulo và logic*

Combinations and permutations are $\binom{n}{r}=n!/[r!(n-r)!]$ and $P(n,r)=n!/(n-r)!$. The greatest common divisor is $\gcd(a,b)$; the least common multiple is $\operatorname{lcm}(a,b)$. Modular exponentiation computes $a^k\bmod m$; a modular inverse $a^{-1}$ satisfies $aa^{-1}\equiv1\pmod m$. The implication $p\Rightarrow q$ is equivalent to $\neg p\lor q$. A counterexample disproves a universal claim with one specific case.

*Tổ hợp và chỉnh hợp lần lượt là $\binom{n}{r}=n!/[r!(n-r)!]$ và $P(n,r)=n!/(n-r)!$. Ước chung lớn nhất là $\gcd(a,b)$; bội chung nhỏ nhất là $\operatorname{lcm}(a,b)$. Lũy thừa modulo tính $a^k\bmod m$; nghịch đảo modulo $a^{-1}$ thỏa $aa^{-1}\equiv1\pmod m$. Mệnh đề kéo theo $p\Rightarrow q$ tương đương $\neg p\lor q$. Một phản ví dụ cụ thể đủ bác bỏ khẳng định tổng quát.*

```python
print('P(5,2), C(5,2) =', math.perm(5, 2), math.comb(5, 2))
print('gcd(48,18), lcm(48,18) =', math.gcd(48, 18), math.lcm(48, 18))
print('3^4 mod 7 =', pow(3, 4, 7), '| nghịch đảo 3 mod 11 =', pow(3, -1, 11))
print('Prime factorization of 360 | Phân tích thừa số nguyên tố của 360 =', sp.factorint(360), '| positive divisors of 12 | ước dương của 12 =', sp.divisors(12))
counterexample = next((n for n in range(1, 10001) if n*n % 12 == 0 and n % 4 != 0), None)
print('Counterexample | Phản ví dụ: n^2 chia hết cho 12 nhưng n không chia hết cho 4:', counterexample)
print('Value of p=>q | Giá trị p=>q với p=True, q=False:', (not True) or False)
```

## 12. Python Data Structures

*Cấu trúc dữ liệu Python*

Lists are mutable; tuples and strings are immutable. In `a[start:stop:step]`, the `stop` index is excluded; `a[::-1]` reverses a sequence. Dictionaries provide methods such as `keys`, `values`, `items`, `get`, `pop`, and `update`; they do not have an `elements` method. A shallow copy shares nested objects, while a deep copy recursively duplicates them.

*Danh sách có thể thay đổi; tuple và chuỗi không thể thay đổi. Trong `a[start:stop:step]`, vị trí `stop` không được lấy; `a[::-1]` đảo thứ tự. Từ điển có các phương thức như `keys`, `values`, `items`, `get`, `pop`, `update`; không có phương thức `elements`. Bản sao nông chia sẻ đối tượng lồng bên trong, còn bản sao sâu sao chép đệ quy.*

```python
items = [10, 20, 30, 40]
print('Slice | Lát cắt:', items[1:3], '| reversed | đảo ngược:', items[::-1])
print('Public methods of dict | Phương thức công khai của dict:', [name for name in dir(dict) if not name.startswith('_')])
original = [[1, 2], [3, 4]]
shallow = original.copy()
import copy
deep = copy.deepcopy(original)
original[0][0] = 99
print('Original | Bản gốc:', original, '| shallow:', shallow, '| deep:', deep)
```
